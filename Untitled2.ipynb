Step 1: Setup and First Look
Goal: Load the data and understand its shape, types, and gaps.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as stats

# Set plotting style and default figure size
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)

# Load the dataset
df = pd.read_csv("train.csv")

# Print shape and preview first 5 rows
print(f"Dataset Shape: {df.shape}")
display(df.head())

# Print information (column names, dtypes, non-null counts)
print("\n--- Dataset Info ---")
df.info()

# Missing value analysis
missing = df.isnull().sum()
missing_pct = (missing / len(df)) * 100
missing_df = pd.DataFrame({'Missing Count': missing, 'Percentage': missing_pct})
missing_df = missing_df[missing_df['Missing Count'] > 0].sort_values(by='Missing Count', ascending=False)

print("\n--- Columns with Missing Values ---")
if not missing_df.empty:
    display(missing_df)
else:
    print("No missing values found!")
```

Step 2: Classify Variables
Goal: Split columns into numerical, categorical, and ordinal groups.

```python
# 1. Identify numeric and categorical columns
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
categorical_cols = df.select_dtypes(exclude=[np.number]).columns.tolist()

# 2. Define ordinal mappings (ordered categories)
ordinal_map = {
    "ExterQual": ["Po", "Fa", "TA", "Gd", "Ex"],
    "KitchenQual": ["Po", "Fa", "TA", "Gd", "Ex"]
}

# Convert specified columns to ordered categorical data types
for col, categories in ordinal_map.items():
    if col in df.columns:
        df[col] = pd.Categorical(df[col], categories=categories, ordered=True)

# 3. Build a summary table
summary_data = []
for col in df.columns:
    # Determine variable type classification
    if col in ordinal_map:
        var_type = "Ordinal"
    elif col in numeric_cols:
        var_type = "Numerical"
    else:
        var_type = "Categorical"

    num_unique = df[col].nunique()
    # Safely obtain non-null example values
    examples = df[col].dropna().unique()[:3].tolist()

    summary_data.append({
        "Column Name": col,
        "Type": var_type,
        "Unique Values": num_unique,
        "Examples": examples
    })

summary_table = pd.DataFrame(summary_data)
display(summary_table)
```

Step 3: Descriptive Statistics
Goal: Summarise centre and spread.

```python
# 1. Numerical Variables Analysis
for column in ['SalePrice', 'GrLivArea', 'LotArea']:
    print(f"=== Descriptive Statistics for {column} ===")

    # Calculate statistics
    col_mean = df[column].mean()
    col_median = df[column].median()
    col_mode = df[column].mode()[0]
    col_std = df[column].std()
    col_var = df[column].var()
    col_range = df[column].max() - df[column].min()

    q75, q25 = np.percentile(df[column].dropna(), [75 ,25])
    col_iqr = q75 - q25

    col_skew = df[column].skew()
    col_kurt = df[column].kurt()

    # Print values
    print(f"Mean: {col_mean:.2f}")
    print(f"Median: {col_median:.2f}")
    print(f"Mode: {col_mode:.2f}")
    print(f"Std Dev: {col_std:.2f}")
    print(f"Variance: {col_var:.2f}")
    print(f"Range: {col_range:.2f}")
    print(f"IQR: {col_iqr:.2f}")
    print(f"Skewness: {col_skew:.2f}")
    print(f"Kurtosis: {col_kurt:.2f}")

    # Determine skewness shape
    if col_mean > col_median:
        print("Distribution Note: Right skewed")
    elif col_mean < col_median:
        print("Distribution Note: Left skewed")
    else:
        print("Distribution Note: Symmetric")
    print("\n")

# 2. Categorical Variables Analysis
for column in ['Neighborhood', 'HouseStyle']:
    print(f"=== Descriptive Statistics for {column} ===")
    mode_val = df[column].mode()[0]
    print(f"Mode: {mode_val}")
    print("Value Counts:")
    print(df[column].value_counts())
    print("\n")
```

Step 5: Correlation and Covariance
Goal: Find numerical features most related to price.

```python
# 1. Select numeric columns and calculate covariance and correlation matrices
numeric_df = df[numeric_cols]
cov_matrix = numeric_df.cov()
corr_matrix = numeric_df.corr()

# 2. Get top 10 features correlated with SalePrice (excluding SalePrice itself)
top_corr = corr_matrix['SalePrice'].abs().sort_values(ascending=False)
top_10_features = top_corr.iloc[1:11] # index 0 is SalePrice itself

print("=== Top 10 Features Correlated with SalePrice ===")
for feature in top_10_features.index:
    r_val = corr_matrix.loc[feature, 'SalePrice']
    print(f"{feature:<15}: Pearson r = {r_val:.4f}")

# 3. For the top 3 features, print covariance and correlation
print("\n=== Top 3 Features: Covariance vs. Correlation ===")
top_3_features = top_10_features.index[:3]
for feature in top_3_features:
    cov_val = cov_matrix.loc[feature, 'SalePrice']
    corr_val = corr_matrix.loc[feature, 'SalePrice']
    print(f"{feature:<15} -> Covariance: {cov_val:14.2f} | Correlation: {corr_val:.4f}")

# 4. Explanation of comparison ease
print("\n=== Why Correlation is Easier to Compare than Covariance ===")
print("1. Scale Dependency: Covariance is unbounded and dependent on the scale of the variables. For example,")
print("   GrLivArea has a covariance in the millions because the values themselves are large (square feet),")
print("   while OverallQual has a covariance in the tens of thousands because it is on a scale of 1 to 10.")
print("2. Standardisation: Correlation standardises the covariance by dividing it by the product of the ")
print("   standard deviations. This constrains correlation to a strict [-1, 1] scale, allowing direct comparison ")
print("   of strength of linear relationships regardless of their units of measurement.")

# 5. Optional: compute Spearman correlation for ordinal columns
print("\n=== Spearman Rank Correlation for Ordinal Columns ===")
for col in ['ExterQual', 'KitchenQual']:
    if col in df.columns:
        # Convert categorical codes to numeric for computation (Po=0, Fa=1, TA=2, Gd=3, Ex=4)
        numerical_ordinal = df[col].cat.codes
        spearman_corr, _ = stats.spearmanr(numerical_ordinal, df['SalePrice'])
        print(f"{col:<15}: Spearman Rank Correlation = {spearman_corr:.4f}")
```

Step 8: Outlier Detection
Goal: Detect outliers with two methods and compare.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Define IQR Outlier Detection Function
def iqr_outliers(series):
    q25 = series.quantile(0.25)
    q75 = series.quantile(0.75)
    iqr = q75 - q25
    lower_bound = q25 - 1.5 * iqr
    upper_bound = q75 + 1.5 * iqr
    return series[(series < lower_bound) | (series > upper_bound)]

# 2. Define Z-score Outlier Detection Function
def zscore_outliers(series, threshold=3):
    mean_val = series.mean()
    std_val = series.std()
    z_scores = (series - mean_val) / std_val
    return series[z_scores.abs() > threshold]

# 3. Analyze and compare outliers for key columns
for col in ['GrLivArea', 'SalePrice']:
    print(f"=== Outlier Detection for {col} ===")
    iqr_flagged = iqr_outliers(df[col])
    zscore_flagged = zscore_outliers(df[col])

    overlap = iqr_flagged.index.intersection(zscore_flagged.index)

    print(f"IQR Method Flagged      : {len(iqr_flagged)} points")
    print(f"Z-score Method Flagged  : {len(zscore_flagged)} points")
    print(f"Overlapping Outliers    : {len(overlap)} points\n")

# 4. Highlight outliers in a Scatter Plot
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='GrLivArea', y='SalePrice', color='gray', alpha=0.5, label='Typical Data')

# Flag outliers with GrLivArea > 4000 and low price (known dangerous outliers)
extreme_outliers = df[(df['GrLivArea'] > 4000) & (df['SalePrice'] < 300000)]
plt.scatter(extreme_outliers['GrLivArea'], extreme_outliers['SalePrice'],
            color='red', edgecolor='black', s=100, zorder=5, label='Extreme Ground-Area Outliers')

plt.title('Outliers in GrLivArea vs SalePrice')
plt.xlabel('GrLivArea (sq ft)')
plt.ylabel('SalePrice ($)')
plt.legend()
plt.show()

# 5. Decide and Clean
# Following standard recommendation (and documented by Ames dataset author Dean De Cock):
# "I would recommend removing any houses with more than 4000 square feet..."
print("=== Decision ===")
print(f"Removing the {len(extreme_outliers)} extreme outliers (GrLivArea > 4000 and SalePrice < $300000) from the dataset.")

df_clean = df.drop(extreme_outliers.index).reset_index(drop=True)
print(f"Original df shape: {df.shape}")
print(f"Cleaned df shape:  {df_clean.shape}")
```

Step 9: Data Quality Checks
Goal: Find duplicates, inconsistent labels, and misleading missing values.

```python
# 1. Check and remove duplicate rows
dup_count = df_clean.duplicated().sum()
print(f"Number of duplicated rows: {dup_count}")
if dup_count > 0:
    df_clean = df_clean.drop_duplicates().reset_index(drop=True)
    print(f"Removed duplicates. New shape: {df_clean.shape}")

# 2. Standardise string columns (strip whitespace, ensure uniform casing)
# We'll work with categorical columns (excluding ordered categorical types like ExterQual)
string_cols = df_clean.select_dtypes(include=['object']).columns
for col in string_cols:
    df_clean[col] = df_clean[col].astype(str).str.strip()

# 3. Handle specific structural features where NaN means "None" (feature absent)
none_fill_cols = ['PoolQC', 'Alley', 'Fence', 'FireplaceQu', 'GarageType']
for col in none_fill_cols:
    if col in df_clean.columns:
        df_clean[col] = df_clean[col].fillna('None')

# 4. Handle numeric columns with missing values (e.g. LotFrontage)
# Strategy: Fill missing LotFrontage values with the median of their Neighborhood
if 'LotFrontage' in df_clean.columns:
    df_clean['LotFrontage'] = df_clean.groupby('Neighborhood')['LotFrontage'].transform(lambda x: x.fillna(x.median()))

# Fallback: Fill any remaining numeric missing values with global median
numeric_cols_clean = df_clean.select_dtypes(include=[np.number]).columns
for col in numeric_cols_clean:
    if df_clean[col].isnull().any():
        df_clean[col] = df_clean[col].fillna(df_clean[col].median())

# 5. Verify remaining missing values in these processed columns
print("\n--- Missing Values Remaining in Key Columns ---")
display(df_clean[none_fill_cols + ['LotFrontage']].isnull().sum())

# 6. Check for impossible temporal values (e.g., YearBuilt > YrSold)
impossible_dates = df_clean[df_clean['YearBuilt'] > df_clean['YrSold']]
print(f"\nHouses with YearBuilt > YrSold: {len(impossible_dates)}")
if not impossible_dates.empty:
    display(impossible_dates[['Id', 'YearBuilt', 'YrSold']])
```
